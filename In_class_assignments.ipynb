```python
sales = float(input("Enter sales ($): "))
tax_rate = float(input("Enter tax rate (%):"))

cogs = sales * 0.4
gross_profit = sales - cogs
taxes = gross_profit * (tax_rate / 100)
net = gross_profit - taxes

print(f"Sales: {sales}$")
print(f"Costs of Goods Sold: {cogs}$")
print(f"Gross Profit: {gross_profit}$")
print(f"Taxes: {taxes}$")
print(f"Net income: {net}$")
```

```python
x = float(input("Enter first number: "))
y = float(input("Enter second number: "))
calc = input("Enter a operator (+,-,/,*,%): ")

if calc == "+":
  print(x + y)
elif calc == "-":
  print(x - y)
elif calc == "/":
  print(x / y)
elif calc == "*":
  print(x * y)
elif calc == "%"
  print(x % y)
elif calc == "**":
  print(x ** y)
elif calc == "//":
  print(x // y)
else:
  print("Invalid Operator")

```

```python
import random
print("Welcome to Rock, Paper, Scissors!")
play = input("Do you want to play rock, paper, scissors? (y/n)")
while play == 'y':
  computer_move = random.choice(['rock','paper','scissors'])
  player_choice = input("What is your choice? ").lower()
  if player_choice != 'rock' and player_choice != 'paper' and player_choice != 'scissors':
    print("Invalid input")
  else:
    if player_choice == computer_move:
      print("It's a tie!")
    elif player_choice == 'rock' and computer_move == 'scissors':
      print("You win!")
    elif player_choice == 'paper' and computer_move == 'rock':
      print("You win!")
    elif player_choice == 'scissors' and computer_move == 'paper':
      print("You win!")
    else:
      print("Computer wins!")

  play = input("Do you want to play ro (y/n) ")

if play == 'n':
  print("Thanks for playing!")
else:
  print("Invalid input")






```
